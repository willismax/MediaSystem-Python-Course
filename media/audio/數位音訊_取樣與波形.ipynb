### 2. 讀取與繪圖

標準實作：讀取 WAV 規格，畫出第一聲道波形。

```python
import wave, numpy as np, matplotlib.pyplot as plt
with wave.open(source_name, 'rb') as w:
    ch, width, rate, frames = w.getnchannels(), w.getsampwidth(), w.getframerate(), w.getnframes()
    data = np.frombuffer(w.readframes(frames), dtype=np.int16).reshape(-1, ch)
print({'channels': ch, 'sample_rate': rate, 'bit_depth': width*8, 'seconds': frames/rate})
t = np.arange(frames)/rate
plt.plot(t, data[:,0], linewidth=.4); plt.xlabel('seconds'); plt.ylabel('amplitude'); plt.show()
```
