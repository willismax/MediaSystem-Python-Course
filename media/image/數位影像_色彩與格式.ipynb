### 1. 安裝與上傳素材

基礎必做：先安裝套件並上傳一張圖片。

```python
!pip -q install pillow
from google.colab import files
from PIL import Image
uploaded = files.upload()
source_name = next(iter(uploaded))
```

### 2. 查看影像資訊

基礎必做：記下尺寸、色彩模式與格式。

```python
with Image.open(source_name) as im:
    print('size:', im.size, 'mode:', im.mode, 'format:', im.format)
    print('EXIF fields:', len(im.getexif()))
```

### 3. 比較尺寸與格式

標準實作：輸出不同尺寸與品質，觀察容量和畫面差異。

```python
from pathlib import Path
out = Path('output'); out.mkdir(exist_ok=True)
with Image.open(source_name) as im:
    im = im.convert('RGB')
    for width in (1200, 800, 400):
        h = round(im.height * width / im.width)
        result = im.resize((width, h))
        result.save(out / f'image_{width}_q90.jpg', quality=90)
        result.save(out / f'image_{width}_q60.jpg', quality=60)
for p in sorted(out.iterdir()): print(p.name, round(p.stat().st_size/1024, 1), 'KB')
```
