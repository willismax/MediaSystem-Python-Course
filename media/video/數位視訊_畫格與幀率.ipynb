### 2. 讀取與擷取

標準實作：顯示尺寸、fps、時長並擷取三個位置。

```python
import cv2, matplotlib.pyplot as plt
cap=cv2.VideoCapture(source_name)
frames=int(cap.get(cv2.CAP_PROP_FRAME_COUNT)); fps=cap.get(cv2.CAP_PROP_FPS)
print({'frames':frames, 'fps':fps, 'seconds':frames/fps})
for ratio in (0, .5, .9):
    cap.set(cv2.CAP_PROP_POS_FRAMES, int(frames*ratio)); ok, frame=cap.read()
    if ok: plt.figure(); plt.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)); plt.axis('off')
cap.release()
```
